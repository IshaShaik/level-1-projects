```python
import serial
import time

# Connect Python to Arduino
arduino = serial.Serial("COM7", 9600)
time.sleep(2)

# Store distance values
distance_list = []

# Function
def check_distance(distance):

    distance_list.append(distance)

    print("--------------------------")
    print("Distance :", distance, "cm")

    if distance < 15:
        print("Status   : STOP")
        print("Buzzer   : ON")
        return "H"

    else:
        print("Status   : SAFE")
        print("Buzzer   : OFF")
        return "L"


# Main Program
while True:

    if arduino.in_waiting > 0:

        data = arduino.readline().decode().strip()

        if data.isdigit():

            distance = int(data)

            command = check_distance(distance)

            arduino.write(command.encode())

    time.sleep(0.1)
```
