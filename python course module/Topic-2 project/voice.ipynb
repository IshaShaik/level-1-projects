```python
import speech_recognition as sr
import webbrowser

r = sr.Recognizer()

pages = {
    "home": "home",
    "about": "about",
    "ai": "ai",
    "contact": "contact"
}

with sr.Microphone(device_index=2) as source:
    print("Speak now...")
    audio = r.listen(source)

command = r.recognize_google(audio).lower()
print("You said:", command)

for page in pages:
    if page in command:
        webbrowser.open(
            "https://ishashaik.github.io/level-1-projects/python%20course%20module/Topic-2%20project/?voice=true#" + page
        )
```
